#### Executing queries

```python
conn = sqlite3.connect(path)

# using a cursor
cursor = conn.execute("SELECT * FROM table;") # a cursor object can be used on results with multiple rows
for row in cursor:
  print(row)
  print(row[0]) # access the first column

# alternative way to use a cursor
cursor = conn.cursor()
cursor.execute("SELECT * FROM table;")
for row in cursor:
  print(row)
cursor.close() # if used in this way, close the cursor as well as the conn when finished

# fetch all results at once
results = conn.execute("SELECT * FROM table;").fetchall() # returns a list of all rows as tuples
print(len(results)) # how many rows were returned

# inserting/deleting data
conn.execute("INSERT INTO table VALUES (value, value, value);")
conn.commit() # commit after making changes to the data
conn.execute("DELETE FROM table WHERE column = value;")
conn.commit()

conn.close()
```
